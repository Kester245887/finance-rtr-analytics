```python
import subprocess

result = subprocess.run(
    ["git", "--version"],
    capture_output=True,
    text=True
)

print(result.stdout)
print(result.stderr)
```

```python
from pathlib import Path
import subprocess

PROJECT_ROOT = Path(
    r"C:\Users\hp\Desktop\Gdansk University of Technology, Poland\Job\AON\Project\finance-rtr-analytics"
)

result = subprocess.run(
    ["git", "status", "--short", "--branch"],
    cwd=PROJECT_ROOT,
    capture_output=True,
    text=True
)

print(result.stdout)
print(result.stderr)
```

```python
for key in ["user.name", "user.email"]:
    result = subprocess.run(
        ["git", "config", "--global", "--get", key],
        capture_output=True,
        text=True
    )
    print(f"{key}: {result.stdout.strip() or 'Not configured'}")
```

```python
import subprocess

result = subprocess.run(
    ["git", "log", "-1", "--stat", "--oneline"],
    cwd=PROJECT_ROOT,
    capture_output=True,
    text=True
)

print(result.stdout)
```
